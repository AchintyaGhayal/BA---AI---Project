```python
import pandas as pd

url = "https://raw.githubusercontent.com/AchintyaGhayal/BA---AI---Project/refs/heads/main/coffee-shop-sales-revenue.csv"

df = pd.read_csv(url, sep="|")

df.head()
```

```python
df.info()
```

```python
print(df.duplicated().sum(), "duplicates")
print(df.isna().sum()) # blanks per column
df = df.drop_duplicates()
df = df.dropna()
print(df.shape) # rows, columns left
```

```python
df.groupby("product_category")["transaction_qty"].sum().plot(
    kind="bar", title="Total Quantity Sold by Product Category")
```
